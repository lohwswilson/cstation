```python
from pw_cli import Client, Env

# This will read the configuration file odoolu.ini
# source_connection = Client.from_config('seq')
source_connection = Client(server='https://skinequality.com', db='PW3-SKNEQY', user='wilson.loh@beyonique.com', password='wai39kua')

# print (source_connection.env['res.users'].browse(source_connection.env.uid))
print (source_connection)

product_template_obj = source_connection.env['product.template']
product_tab_line_obj = source_connection.env['product.tab.line']

products = product_template_obj.search([(1,'=',1)])

for line in products:
    is_how_to_use = False
    is_ingredients = False
    # Search Product Tab for the product ID
    product_tabs = product_tab_line_obj.search([('product_id','=', line.id)])
    # Check whether is there any product tab
    if len(product_tabs) == 0:
        is_ingredients = False
        is_how_to_use = False
    else:
        print('Check Product Tab: %s' % line.name)  
        for rec in product_tabs:
            if rec.tab_name == 'Directions':
                rec.tab_name = 'How To Use'
                print('CHANGE Directions --> How To Use')
            if rec.tab_name == 'How To Use':
                is_how_to_use = True
            if rec.tab_name == 'Ingredients':
                is_ingredients = True
    # Create Product Tabs
    if not is_how_to_use:
        print('Create How To Use Tab for : %s' % line.name)
        vals = {
            'product_id': line.id,
            'tab_name' : 'How To Use'
        }
        product_tab_line_obj.create(vals)    
    if not is_ingredients:
        print('Create Ingredients Tab for : %s' % line.name)
        vals = {
            'product_id': line.id,
            'tab_name' : 'Ingredients'
        }
        product_tab_line_obj.create(vals)  
```
